Data Cleaning and Preparation

```python
# to check missing values
print("Missing values before cleaning:\n", df.isnull().sum()[df.isnull().sum() > 0])
```

```python
#grouping the rows by category, finds the groups like season or crop
df['Rainfall_mm'] = df.groupby('Season')['Rainfall_mm'].transform(lambda x: x.fillna(x.median()))
```

```python
#calculating the median value for each group
df['Soil_Moisture_pct'] = df.groupby('Season')['Soil_Moisture_pct'].transform(lambda x: x.fillna(x.median()))
```

```python
#replacing the missing value with nan and by calculating the median value of that specific group
df['Yield_Tonnes_Ha'] = df.groupby('Crop')['Yield_Tonnes_Ha'].transform(lambda x: x.fillna(x.median()))
```

```python
# Verify no remaining missing values
print("\nMissing values after cleaning:\n", df.isnull().sum().sum())
```

Exploratory Data Analysis (EDA) & Visualizations

1. Seasonal Yield & Production Comparison

```python
# by season we are plotting the average yield
plt.figure(figsize=(8, 4))
sns.barplot(data=df, x='Season', y='Yield_Tonnes_Ha', estimator=np.mean, errorbar=None, palette='Blues_d')
plt.title('Average Crop Yield (Tonnes/Ha) by Season')
plt.ylabel('Yield (Tonnes/Ha)')
plt.show()
```

2. Seasonal Financial & Profit Analysis

```python
# we are plotting total revenue vs total cost by season
df_financials = df.groupby('Season')[['Total_Cost_INR', 'Revenue_INR', 'Profit_INR']].mean().reset_index()

df_financials.plot(x='Season', y=['Total_Cost_INR', 'Revenue_INR', 'Profit_INR'], kind='bar', figsize=(10, 5))
plt.title('Average Financial Performance by Season')
plt.ylabel('Amount (INR)')
plt.xticks(rotation=0)
plt.show()
```

3. Environmental Conditions vs. Resource Usage

```python
# Scatter plot Rainfall vs Yield and is colored by Season
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='Rainfall_mm', y='Yield_Tonnes_Ha', hue='Season', alpha=0.7)
plt.title('Impact of Rainfall on Crop Yield Across Seasons')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Yield (Tonnes/Ha)')
plt.show()
```

4. Pest Risk & Water Efficiency Analysis

```python
# Boxplot of Pest Risk by Season
plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x='Season', y='Disease_Pest_Risk_pct', palette='Set2')
plt.title('Disease & Pest Risk (%) Across Seasons')
plt.show()
```
